#Series

```python
import pandas as pd

x = [1,2,3,4,7,9]
var = pd.Series(x , index = ['q','w','e','r','t','y'] , dtype= "float", name= "python")

print(var)
print(type(var))
print(var[4])
```

```python
dic = {"name":["yash","raj","kyar","viraj"], "por": [12,13,14,15,26,34], "rank": [1,2,3,4,5,6]}
var1 = pd.Series(dic)
print(var1)
print(type(var1))
```

```python
s = pd.Series(13 , index = ['q','w','e','r'])
print(s)
print(type(s))
```

```python
s1 = pd.Series(14 , index = ['q','w','e','r'] , dtype = 'int') 
s2 = pd.Series(13 , index = ['q','w'] , dtype = 'int')

#result = (s1 + s2).fillna(5).sort_index() ## .fillna() method to fill null values
result = (s1 + s2).fillna(5).reindex(['q', 'w', 'e', 'r']) ## .fillna() method to fill null values
print(result)
print(type(s1 + s2))
#s1 + s2

## .sort_index() method sort alphabetically
## to reorder indices i have to use .reindex([1,2,/'a','c'])

```

#Some Methods to handle NaN

Method	Description
.fillna()	Replace NaN with a specific value.
.dropna()	Remove rows or columns containing NaN.
.isna()	Return a boolean mask where NaN is True.
.notna()	Return a boolean mask where non-NaN is True.

```python
s = pd.Series([1 , 3 , None , 4])
#s
s.fillna(5)
```

```python
s = pd.Series([1 , 3 , None , 4])
#s
s.dropna()
```

```python
s = pd.Series([1 , 3 , None , 4])
#s
s.isna()
```

```python
s = pd.Series([1 , 3 , None , 4])
#s
s.notna()
```
