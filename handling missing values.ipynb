```python
var_2 = pd.read_csv("C:\\Users\\asus\\Documents\\pandas_unleashed\\loan.csv")
var_2.head(15)
```

```python
#fillna

fill_null = var_2.fillna("YASH")
fill_null.head(17)
```

```python
rows_with_nulls = var_2[var_2.isnull().any(axis = 1)]
rows_with_nulls.head(10)

cols_with_nulls = var_2.columns[var_2.isnull().any(axis = 0)]
cols_with_nulls
```

```python
# filling particular value in any column
fill_in_col = var_2.fillna({"Gender": "Hello"}) 
fill_in_col.head(4)
```

```python
#forward and backward filling
#frwd = var_2.fillna(method = "ffill")
#frwd.head(20)

bckwd = var_2.fillna(method = "bfill")
bckwd.head(4)
```

```python
#finding those particulars in which null values are filled

#get_nulls = var_2[var_2.isnull().any(axis = 1)]
#get_nulls

# Fill null values with "YASHRAJPUT"
filled_var_2 = var_2.fillna("YASHRAJPUT")

# Find locations where null values were filled
filled_locations = var_2.isnull() & ~filled_var_2.isnull()

# Identify rows where null values were filled
rows_with_filled_nulls = filled_locations.any(axis=1)

# Display rows where null values were filled
print("Rows where null values were filled:")
print(filled_var_2[rows_with_filled_nulls])
print(filled_var_2[rows_with_nulls])
```

```python

columns_with_filled_nulls = filled_locations.any(axis=0)
print("Columns where null values were filled:")
print(filled_var_2.loc[:, columns_with_filled_nulls])


filled_null_counts = filled_locations.sum(axis=0)
print("Count of filled null values in each column:")
print(filled_null_counts)
```
