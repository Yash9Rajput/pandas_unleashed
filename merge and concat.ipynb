#merge and concat

```python
import pandas as pd
```

```python
var1 = pd.DataFrame({"A":[1,2,3,4] , "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5] , "B":[9,10,11,12]})
#result = pd.merge(var1 , var2)
#result = pd.merge(var1 , var2 , on="A")
#result = pd.merge(var1 , var2 , how="outer" , indicator = True)
#result = pd.merge(var1 , var2 , left_index= True , right_index= True)
#result = pd.merge(var1 , var2 , left_index= True , right_index= True)
result = pd.merge(var1 , var2 , left_index= True , right_index= True , suffixes=("name","java"))
result
```

#concat

```python
#concating two series
sr1 = pd.Series({"A": [1,2,3,4]})
sr2 = pd.Series({"B": [11,12,13,14]})
pd.concat([sr1 , sr2],axis = 0)
```

```python
#concating two dataframes
d1 = pd.DataFrame({"A":[1,2,3,4] , "B":[11,12,13,14]})
d2 = pd.DataFrame({"A":[5,6,7,8] , "C":[55,56,57,58]})
pd.concat([d1 , d2] ,axis = 0)
```

```python
#concating two dataframes
d1 = pd.DataFrame({"A":[1,2,3,4] , "B":[11,12,13,14]})
d2 = pd.DataFrame({"A":[5,6,7,8] , "C":[55,56,57,58]})
#pd.concat([d1 , d2] ,axis = 1, join="outer")
pd.concat([d1 , d2] ,axis = 0, keys=["d1","d2"])
```

```python
#concating two dataframes
d1 = pd.DataFrame({"D":[1,2,3,4]})
d2 = pd.DataFrame({"A":[5,6,7,8] , "C":[55,56,57,58]})
pd.concat([d1,d2] , axis = 1 , keys=["d1" , "d2"])
```
