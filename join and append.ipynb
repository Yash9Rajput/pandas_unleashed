#join and append

```python
import pandas as pd

var1 = pd.DataFrame({"X": [1, 2, 3], "Y": [4, 5, 6]} , index=["a","b","c"])
var2 = pd.DataFrame({"Z": [7, 8], "Y": [10, 11]} , index=["a","b"])
var2.join((var1) ,how="right" , lsuffix = "_yash" , rsuffix="_raj")
```

```python
var1 = pd.DataFrame({"X": [1, 2, 3], "Y": [4, 5, 6]})
var2 = pd.DataFrame({"Z": [7, 8], "Y": [10, 11]})
result = var1.append(var2, ignore_index=True)
result
```
