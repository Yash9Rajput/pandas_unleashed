#melt

```python
import pandas as pd

var = pd.DataFrame({"days":[1,2,3,4,5,6] , 
                    "eng":[10, 20, 30, 40, 50, 60] , 
                    "maths":[17, 18, 19, 20, 21, 22]
                })

var
```

```python
melted = pd.melt(var , id_vars=["days"], var_name="subjects", value_name="scores")
#melted
result = melted.pivot(index = "days", columns="subjects", values="scores").reset_index()

result.columns.name = None

result
```

#pivot

```python
import pandas as pd

var = pd.DataFrame({"days":[1,2,3,4,5,6] ,
                    "st_name": ["a","a","b","b","a","b"] ,
                    "eng":[10, 20, 30, 40, 50, 60] , 
                    "maths":[17, 18, 19, 20, 21, 22]
                })

result = var.pivot(index = "days" , columns="st_name", values=["eng" , "maths"])
final = result.fillna("000000")
final
```

```python
var = pd.DataFrame({"days":[1,1,1,2,2,2] ,
                    "st_name": ["a","a","b","b","a","b"] ,
                    "eng":[10, 20, 30, 40, 50, 60] , 
                    "maths":[17, 18, 19, 20, 21, 22]
                })

result = var.pivot_table(index = "st_name" , columns="days", aggfunc="mean" , margins = True)
#final = result.fillna(0.0)
# final
result
```
