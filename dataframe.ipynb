##Dataframe

```python
import pandas as pd

l = [1,2,3,4,5]
var = pd.DataFrame(l)
print(var)
```

```python
dic = {'a':[1,2,3] , 'b':[4,5,None] , 'c':[1,3,4] , 1:['a','f','g']}
#var = pd.DataFrame(dic).fillna(9)
var = pd.DataFrame(dic , columns = ["a" , 1] , index = ['q','w','e'])
print(var)
print(var['a'][2])
print(type(var))


d = {'a':[1,2,3] , 'b':[4,None,None]}
var = pd.DataFrame(d).fillna(7)
print(var)
```

```python
#creating a dataframe using two lists

list_1 = [[1,2,3,4,5] , ['q','w','e','r','t']]
var_2 = pd.DataFrame(list_1 , index= ['A' , 'B'])
print(type(var_2))
print(var_2)
```

```python
#Creating a DataFrame using two series

sr = {"s" : pd.Series([1,2,3]) , "r": pd.Series([4,5,6])}
var_3 = pd.DataFrame(sr)
print(var_3)
```
