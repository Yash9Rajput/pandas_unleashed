#group by

```python
import pandas as pd

var = pd.DataFrame({"Name": ["a","b","c","a","a","d","e","b","c","a","e","d"] , 
                    "S_1": [1, 2, 3, 1, 1, 5, 3, 3, 1, 4, 3, 2] , 
                    "S_2": [6, 8, 7, 8, 5, 5, 9, 7, 10,10, 7, 6]})

var
```

```python
var_new = var.groupby("Name")
var_new
```

```python
for x,y in var_new:
    print(x)
    print(y)
    print()
```

```python
var_new.get_group("c")
```

```python
var_new.min()
```

```python
var_new.max()
```

```python
var_new.mean()
```
